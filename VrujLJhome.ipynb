# Armstrong Number

```python
n = int(input("Enter number: "))

temp = n
l = 0
arms = 0

# Count digits
while temp > 0:
    l += 1
    temp //= 10

print(l)

# Calculate Armstrong sum
temp = n

while temp != 0:
    digit = temp % 10
    arms = arms + (digit ** l)
    temp = temp // 10

if n == arms:
    print(n, "is  armstrong number.")
else:
    print(n, "is not  armstrong number.")
```

# Disarium number


```python
n = int(input("Enter number: "))

temp = n
l = 0
d = 0

while temp > 0:
    l += 1
    temp //= 10


temp = n

for i in range(l, 0, -1):
    digit = temp % 10
    d = d + (digit ** i)
    temp //= 10

print("Sum:", d)

if n == d:
    print(n, "is  disarium number.")
else:
    print(n, "is not disarium number.")
```
